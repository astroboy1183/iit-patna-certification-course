# Best Practices — Text Splitting for RAG

###  Goal
Keep chunks semantically meaningful *and* model-friendly.

---

##  Recommended Settings

| Splitter Type | When to Use | Typical Chunk Size | Overlap |
|----------------|-------------|--------------------|----------|
| `RecursiveCharacterTextSplitter` | General text, PDFs | 500–1000 chars | 50–100 |
| `TokenTextSplitter` | Token-aware LLMs (OpenAI, Gemini) | 200–400 tokens | 20–40 |

---

##  Practical Tips

1️⃣ Always ensure **chunks are context-complete** (avoid cutting mid-sentence).  
2️⃣ **More overlap = more redundancy = higher retrieval cost**, so balance it.  
3️⃣ Use **token-based splitting** for multilingual or dense technical text.  
4️⃣ Keep **chunk size < 10–15% of model context** to avoid truncation.  
5️⃣ Save your chunks — embedding them repeatedly is wasteful.

---

##  Quick Analogy
> Think of splitting like slicing a pizza:  
> Too many small slices → more work for the model;  
> Too few → hard to eat properly.  
> The goal is to find the sweet spot 🍕.
