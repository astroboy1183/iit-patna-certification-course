# Retriever Comparison Summary

Retrievers are the **core of any RAG (Retrieval-Augmented Generation)** system.  
They decide *what context* gets passed to the LLM — making them the backbone of response quality.

This section introduced **three levels of retrieval intelligence** in LangChain:

---

## 1️⃣ BaseRetriever — The Foundation
**Concept:** Fetches top-k documents that are semantically similar to the query.

### Example
**Query:** “How does LangChain help developers?”
- ✅ Retrieves documents most similar in meaning.
- ❌ Might miss relevant info if phrasing differs.

### Key Properties
| Aspect | Description |
|--------|--------------|
| Retrieval Type | Basic similarity search |
| Data Source | Any vector store (FAISS, Chroma, Pinecone, etc.) |
| Strength | Simple, fast, effective |
| Limitation | Recall limited to single query phrasing |

### Use When
- Dataset is clean and concise.  
- You need quick local search (e.g., FAISS demo).  
- Recall (coverage) is already good.

---

## 2️⃣ MultiQueryRetriever — Boost Recall
**Concept:** Uses an LLM to **rephrase the user query multiple times**, runs all variants, and merges results.

### Example
**Original Query:** “How does LangChain retrieve relevant data?”
**Generated Variants:**
- “How does LangChain find information?”
- “Explain how LangChain performs retrieval.”
- “What mechanism does LangChain use for data fetching?”

The retriever runs all these queries → merges results → gives broader coverage.

### Key Properties
| Aspect | Description |
|--------|--------------|
| Retrieval Type | Multi-query similarity |
| Powered By | LLM (e.g., GPT-3.5) for paraphrasing |
| Strength | Improves recall — catches missed documents |
| Limitation | Slightly slower, requires LLM calls |

### Use When
- Queries can be ambiguous or phrased differently.  
- Dataset is large/diverse in writing style.  
- You want **higher recall** (coverage of relevant docs).

---

## 3️⃣ ContextualCompressionRetriever — Boost Precision
**Concept:** Fetches top-k docs, then uses an LLM to **filter and condense** them, keeping only query-relevant content.

### Example
**Query:** “What are retrievers used for in LangChain?”
- Base retriever may return full paragraphs.
- Contextual retriever filters those paragraphs to just:
  > “Retrievers fetch relevant chunks of information to answer a query.”

### Key Properties
| Aspect | Description |
|--------|--------------|
| Retrieval Type | Post-retrieval filtering |
| Powered By | LLM-based document compressor |
| Strength | High precision — only relevant info retained |
| Limitation | Slightly higher latency and cost (extra LLM step) |

### Use When
- Documents are **long or noisy**.  
- You need **focused, concise** context for the LLM.  
- Reducing token count matters (e.g., paid APIs).

---

## Summary Comparison

| Feature | BaseRetriever | MultiQueryRetriever | ContextualRetriever |
|----------|----------------|---------------------|----------------------|
| **Goal** | Fast semantic search | Improve recall | Improve precision |
| **Uses LLM?** | ❌ | ✅ (for query expansion) | ✅ (for context filtering) |
| **Speed** | ⚡ Fast | ⚙️ Moderate | 🧩 Moderate |
| **Accuracy Type** | Good recall | High recall | High precision |
| **Ideal For** | Small datasets | Large, diverse corpora | Long or messy documents |

---

## Practical Analogy

| Retriever | Real-world Analogy |
|------------|--------------------|
| **BaseRetriever** | You Google exactly one search phrase |
| **MultiQueryRetriever** | You Google 5 reworded phrases and merge results |
| **ContextualRetriever** | You summarize only the best snippets from those pages |

---

## Final Guidance

| Scenario | Best Retriever |
|-----------|----------------|
| Small, well-curated dataset | BaseRetriever |
| Diverse text with mixed phrasing | MultiQueryRetriever |
| Long documents, noisy content | ContextualCompressionRetriever |

---

**Pro Tip:**  
In production RAG systems, it’s common to *chain* them:
> MultiQueryRetriever → ContextualCompressionRetriever  
for both high recall **and** high precision.
