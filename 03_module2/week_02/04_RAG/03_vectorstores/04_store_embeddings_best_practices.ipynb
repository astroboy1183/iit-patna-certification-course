# 📘 Best Practices — Storing Embeddings

### Goal
Understand how to store and manage vector embeddings efficiently.

---

## Recommended Approaches

| Use Case | Recommended Vector Store |
|-----------|---------------------------|
| Local RAG prototypes | FAISS |
| Persistent, reloadable RAG apps | Chroma |
| Cloud-scale applications | Pinecone / Weaviate / Milvus |

---

## Tips

1️⃣ **Cache embeddings** — never re-embed the same text.  
2️⃣ **Use local SentenceTransformer models** for CPU demos.  
3️⃣ **FAISS** = Speed ⚡, **Chroma** = Simplicity 💾  
4️⃣ Keep embedding dimensions consistent across runs.  
5️⃣ Save index files alongside metadata for easy reuse.

---

## Analogy
> Think of FAISS as your **RAM notebook** — quick but temporary.  
> Think of Chroma as your **hard drive notebook** — slower but persistent.
